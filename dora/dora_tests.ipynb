# Badge 3: DORA checks

Every DORA step from the course, in order. Each check has two cells:

1. **Local check**: the course's check query with the external `GRADER` call removed. It runs entirely in this account and sends nothing.
2. **DORA (exact course code)**: verbatim and unedited (DORA audits a hash of the code).

⚠️ **Running a DORA cell sends data outside this account** to Snowflake Education's grading service. Run only the local checks unless you intend to submit.

Each check is meant to be run at the end of its lesson: later lessons rewrite the objects earlier checks read (for example ORDERS). DABW005 reads `SNOWFLAKE.ACCOUNT_USAGE`, which can lag by up to a few hours.
