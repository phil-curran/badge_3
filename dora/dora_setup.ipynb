⚠️ The cells create the API integration and the GRADER function that call Snowflake Education's external grading service. They are shared with the other badges; if another badge's DORA setup already ran, these cells recreate the same objects.

The course page only says to follow the DORA tutorial (create UTIL_DB, the API integration and the GRADER function). The script below is the same one the other badge courses print (Badge 5 and Badge 4 Lesson 1); it is not in this course's export. The `DORA_IS_WORKING` test that calls the grader is in `dora/dora_tests.ipynb`.

### 1. Run the API Integration Script

```
%%sql -r dora_api_integration
use role accountadmin;

create or replace api integration dora_api_integration
api_provider = aws_api_gateway
api_aws_role_arn = 'arn:aws:iam::321463406630:role/snowflakeLearnerAssumedRole'
enabled = true
api_allowed_prefixes = ('https://awy6hshxy4.execute-api.us-west-2.amazonaws.com/dev/edu_dora');
```

### 2. Set up the Grader Function

```
%%sql -r dora_grader_function
create or replace external function util_db.public.grader(
      step varchar
    , passed boolean
    , actual integer
    , expected integer
    , description varchar)
returns variant
api_integration = dora_api_integration 
context_headers = (current_timestamp, current_account, current_statement, current_account_name)
as 'https://awy6hshxy4.execute-api.us-west-2.amazonaws.com/dev/edu_dora/grader'
;
```
